```python
def show_bounded(img, title=""):

    plt.imshow(img, vmax=1, vmin=0, cmap="gray")
    plt.colorbar()
    plt.title(title)
    plt.show()

    return

def show(img, title=""):

    plt.imshow(img, cmap="gray")
    plt.colorbar()
    plt.title(title)
    plt.show()

    return
```

```python
img_big = cv2.imread("test_img/big_noisy_center_square.png")
img_small = cv2.imread("test_img/center_square.png")

mask1  = cv2.cvtColor(img_big, cv2.COLOR_BGR2GRAY)
mask2 = cv2.cvtColor(img_small, cv2.COLOR_BGR2GRAY)

show(mask1, "big objects")
show(mask2, "small object")

ok, componant = get_overlap_componant(mask1, mask2)

if ok:
    show(componant)
else:
    print("Nothing")
```
